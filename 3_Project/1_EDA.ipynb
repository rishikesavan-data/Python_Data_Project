# Exploratory Data Analysis

### Roles to Explore :

```python
# Importing Libraries
import ast
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datasets import load_dataset
from matplotlib.ticker import FuncFormatter
from adjustText import adjust_text
```

```python
# Loading Data 
raw = load_dataset('lukebarousse/data_jobs')
df = raw['train'].to_pandas().copy()

# Data Cleaning
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
df
```

### Filter for US and India Data Analyst Role:

```python
df_us =df[(df['job_country']=='United States') & (df['job_title_short']=='Data Analyst')]
df_loc =df_us['job_location'].value_counts().head(10).to_frame()

sns.set_theme(style='ticks')
sns.barplot(data=df_loc,x='count',y=df_loc.index,hue='count',palette='dark:b_r',legend=False)
sns.despine()

plt.grid(axis='x',alpha =0.3,linestyle='--')
plt.ylabel('')
plt.xlabel('Number of Jobs')
plt.title('Counts of Job locations for Data Analyst in the US and India')
plt.show()
```

### Job Opportunity

```python
dicts = {'job_work_from_home':'Work From Home',
'job_no_degree_mention':'No Degree Mention',
'job_health_insurance':'Health_Insurance'}

fig ,ax = plt.subplots(1,3,figsize=(12,4))

for i,(column,titles) in enumerate(dicts.items()):
    ax[i].pie(df_us[column].value_counts(),labels=['False','True'],startangle=90,autopct='%1.1f%%')
    ax[i].set_title(titles,fontweight='bold',fontsize=15)
```

### Companies to Explore

```python
df_comp =df_us['company_name'].value_counts().head(10)
df_comp =df_comp.to_frame()
df_comp
```

```python
sns.barplot(data=df_comp,x='count',y=df_comp.index,hue='count',palette='dark:b_r',legend=False)
plt.ylabel('')
plt.xlabel('Number of Jobs')
plt.title('Count of Companies for Data Analyst in the US')
sns.despine()
plt.show()
```
