# Matplotlib Labelling

**Methods**

- plt.title()
- plt.ylabel() or plt.xlabel()
- plt.xticks() or plt.yticks()

```python
from datasets import load_dataset
import pandas as pd
import matplotlib.pyplot as plt

raw = load_dataset('lukebarousse/data_jobs') 
df = raw.copy()
df =df['train'].to_pandas()
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
df['job_posted_month'] = df.job_posted_date.dt.month
df.info()
```

```python
monthly_job_count =df['job_posted_month']
monthly_count =monthly_job_count.value_counts().sort_index()
```

```python
monthly_count
```

- plt.title()
- plt.xlabel() / plt.ylabel()

```python
plt.plot(monthly_count.index,monthly_count)
# In the below graph, we can't able to figure out what is this chart for ?

# That's why we need a title for this graph
plt.title("Monthly Job Count")

# but still, we don't know what the axis are representing. so we need to fix it
plt.ylabel("Job Count") # y axis
plt.xlabel("Month") # x axis
```

- plt.xtick()

```python
# in the previous section we came across a title overlapping problem in bar chart. now we are going to solve that
jobs =df['job_title_short'].value_counts()
jobs
```

```python
plt.bar(jobs.index,jobs)
plt.xticks(rotation = 45) # now there is a new problem the title is not aligned properly
plt.show()
```

```python
# to solve this alignment issue i use ha = 'right' inside xticks() method 
plt.bar(jobs.index,jobs)
plt.xticks(rotation = 45, ha ='right')
plt.show()
```
