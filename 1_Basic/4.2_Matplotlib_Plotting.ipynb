# Matplotlib Plotting

**Methods**

- plt.plot()

- plt.show()
- value_count()
- sort_index()
- sort_values()


```python
from datasets import load_dataset
import pandas as pd
import matplotlib.pyplot as plt

raw = load_dataset('lukebarousse/data_jobs') 
df = raw.copy()
df =df['train'].to_pandas()
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
df.info()
```

```python
x=[1,2,3,4,5]
y=[1,2,3,4,5]
plt.plot(x,y)
```

```python
df['job_posted_month'] = df.job_posted_date.dt.month
df.info()
df
```

**line chart/ Trend Chart**

```python
monthly_count =df['job_posted_month'].value_counts().sort_index()
monthly_count
```

```python
plt.plot(monthly_count.index,monthly_count)
plt.show() # This line remove this things [<matplotlib.lines.Line2D at 0x1e45c5da300>]
```

**Bar Chart**

```python
jobs =df['job_title_short'].value_counts()
jobs
```

```python
plt.bar(jobs.index,jobs) # This chart cause us some problem see the x-axis job_titles are not listed properly its overlap.
# That's why we are going to use horizontal bar chart
```

**Horizontal Bar Chart**

```python

plt.barh(jobs.index,jobs)
# look this show the values in ascending order, for to fix that we use to sort them.
```

```python
sorted_jobs =jobs.sort_values(ascending=True)
sorted_jobs
```

```python
plt.barh(sorted_jobs.index,sorted_jobs)
```
