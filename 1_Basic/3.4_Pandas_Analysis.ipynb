# Pandas : Analysis

**Methods:**

- count()
- sum()
- cumsum()
- min()
- max()
- idxmin()
- idxmax()
- mean()
- median()
- mode()
- value_counts()
- groupby()

```python
from datasets import load_dataset
import pandas as pd

dataset = load_dataset('lukebarousse/data_jobs')
raw=dataset['train'].to_pandas()
df = None
df = raw.copy()
# Dataset 
df['job_posted_date']=pd.to_datetime(df.job_posted_date)
df.info()
```

**count() Method**


```python
df.count()
```

**sum() Method**

```python
df['salary_year_avg'].sum()
```

**cumsum() Method**

```python
df['salary_year_avg'].notna().cumsum()
```

**min() Method**

```python
df['salary_year_avg'].min()
```

**max() Method**

```python
df['salary_year_avg'].max()
```

**idxmin() Method**

```python
df['salary_year_avg'].idxmin() 
```

```python
min_salary =df['salary_year_avg'].idxmin()
df.iloc[min_salary]
```

**idxmax() Method**

```python
df['salary_year_avg'].idxmax()
```

```python
max_salary = df['salary_year_avg'].idxmax()
df.iloc[max_salary]
```

**mean() Method**

```python
df['salary_year_avg'].mean()
```

**median() Method**

```python
df['salary_year_avg'].median()
```

**mode() Method**

```python
df['salary_year_avg'].mode()
```

**value_counts() Method**

The value_counts() method finds all unique items in a Pandas column and counts exactly how many times each one appears. It then automatically displays these items as a sorted list, placing the most frequent values at the very top.

```python
df['job_title_short'].value_counts() # it count the total unique 
```

**groupby() Method**

```python
df.groupby('job_title_short')['salary_year_avg'].min()
```

```python
df.groupby(['job_title_short','job_country'])['salary_year_avg'].max()
```

```python
df.groupby('job_title_short')[['salary_year_avg','salary_hour_avg']].mean()
```

```python
df.groupby('job_title_short')['salary_year_avg'].agg(['min','max','mean','median','count'])
```

```python
df.groupby('job_title_short')['salary_year_avg'].agg(['min','max','mean','median','count']).sort_values('count')
```
