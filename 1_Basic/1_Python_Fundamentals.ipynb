### 03_Strings (Important)

**Document**
- Indexing
- Slicing
- String methods

**Important Methods**
- upper()
- lower()
- strip()
- replace()
- split()
- join()
- find()
- count()

**Data Analyst Use**

- Data cleaning

```python

print(string) # This will print the value of the variable 'string', which is "Hello, Python!"
print(string.upper()) # This will print the uppercase version of the string, which is "HELLO, PYTHON!"
print(string.lower()) # This will print the lowercase version of the string, which is "hello, python!"
print(string.strip()) # This will remove any leading and trailing whitespace from the string, resulting in "Hello, Python!"
print(string.replace("Python", "World")) # This will replace "Python" with "World" in the string, resulting in "Hello, World!"
print(string.split()) # This will split the string into a list of words, resulting in ['Hello,', 'Python!']
print(string.find("Python")) # This will return the index of the first occurrence of "Python" in the string, which is 7
print(string.count("o")) # This will return the number of occurrences of "o" in the string, which is 2
print(len(string)) # This will return the length of the string, which is 13
print(string.index("Python")) # This will return the index of the first occurrence of "Python" in the string, which is 7
print(string.capitalize()) # This will capitalize the first character of the string, resulting in "Hello, python!"
print(string.title()) # This will capitalize the first character of each word in the string, resulting in "Hello, Python!"
print(string.swapcase()) # This will swap the case of each character in the string, resulting in "hELLO, pYTHON!"
print(string.center(20)) # This will center the string within a field of width 20, resulting in "   Hello, Python!   "

print(string.ljust(20)) # This will left-justify the string within a field of width 20, resulting in "Hello, Python!       "
print(string.rjust(20)) # This will right-justify the string within a field of width 20, resulting in "       Hello, Python!"
print(string.startswith("Hello")) # This will return True if the string starts with "Hello", otherwise False
print(string.endswith("Python!")) # This will return True if the string ends with "Python!", otherwise False
print(string.zfill(20)) # This will pad the string with zeros on the left to make it 20 characters long, resulting in "000000Hello, Python!"

print(string.isprintable()) # This will return True if all characters in the string are printable, otherwise False
print(string.islower()) # This will return True if all characters in the string are lowercase, otherwise False
print(string.isupper()) # This will return True if all characters in the string are uppercase, otherwise False
print(string.isspace()) # This will return True if all characters in the string are whitespace, otherwise False
print(string.isalpha()) # This will return True if all characters in the string are alphabetic, otherwise False
print(string.isdigit()) # This will return True if all characters in the string are digits, otherwise False
print(string.isalnum()) # This will return True if all characters in the string are alphanumeric, otherwise False
```

**Document**

*Arithmetic:*

```python
# arithmetic operators
a = 10
b = 5

print('Arithmetic Operators:')
print(a + b)  # Addition, will print 15
print(a - b)  # Subtraction, will print 5
print(a * b)  # Multiplication, will print 50
print(a / b)  # Division, will print 2.0
print(a // b) # Floor division, will print 2
print(a % b)  # Modulus, will print 0
print(a ** b) # Exponentiation, will print 100000

# comparison operators
print('\nComparison Operators:')
print(a == b)  # Equal to, will print False
print(a != b)  # Not equal to, will print True
print(a > b)   # Greater than, will print True
print(a < b)   # Less than, will print False
print(a >= b)  # Greater than or equal to, will print True
print(a <= b)  # Less than or equal to, will print False

# logical operators
print('\nLogical Operators:')
print(a > 5 and b < 10)  # Logical AND, will print True
print(a > 5 or b < 3)    # Logical OR, will print True
print(not (a > 5))       # Logical NOT, will print False
print(a > 5 and b < 3)   # Logical AND, will print False
print(a > 5 or b < 3)    # Logical OR, will print True
print(not (a > 5 and b < 3)) # Logical NOT, will print True

# assignment operators
print('\nAssignment Operators:')
c = a  # Assignment, c will be 10
c += b # Addition assignment, c will be 15
c -= b # Subtraction assignment, c will be 10
c *= b # Multiplication assignment, c will be 50
c /= b # Division assignment, c will be 10.0
c //= b # Floor division assignment, c will be 2.0
c %= b  # Modulus assignment, c will be 0.0
c **= b # Exponentiation assignment, c will be 0.0

# bitwise operators
print('\nBitwise Operators:')
print(a & b)  # Bitwise AND, will print 0
print(a | b)  # Bitwise OR, will print 15
print(a ^ b)  # Bitwise XOR, will print 15
print(~a)     # Bitwise NOT, will print -11
print(a << 1) # Bitwise left shift, will print 20
print(a >> 1) # Bitwise right shift, will print 5


```

```python
sales = 1000

# If statement examples
print("if Statement Example:")
if sales > 500:
    print("Sales are good!")

# If-else statement examples
print("\nif-else Statement Example:")
if sales > 1500:
    print("Sales are good!")
else:
    print("Sales are not good!")

# If-elif-else statement examples
print("\nif-elif-else Statement Example:")
if sales > 1500:
    print("Sales are good!")
elif sales > 500:
    print("Sales are average!")
else:
    print("Sales are not good!")

# Nested if statement examples
print("\nNested if Statement Example:")
if sales > 500:
    if sales > 1000:
        print("Sales are excellent!")
    else:
        print("Sales are good!")
```

### 07_lists (Important)

Very important for Data Analysis.

**Methods:**

- append()
- extend()
- insert()
- remove()
- pop()
- sort()
- reverse()
- copy()

Using id() Method to know the memory address

```python

list1 = [1, 28, 33, 4, 3,2,78]

print("\nappend(6) method example:")
list1.append(6)  # This will add the number 6 to the end of the list, resulting in [1, 28, 33, 4, 3, 6]
print(list1)  # This will print the updated list, which is [1, 28, 33, 4, 3, 6]

print("\nextend([7, 8, 9]) method example:")
list1.extend([7, 8, 9])  # This will add the numbers 7, 8, and 9 to the end of the list, resulting in [1, 28, 33, 4, 3, 6, 7, 8, 9]
print(list1)  # This will print the updated list, which is [1, 28, 33, 4, 3, 6, 7, 8, 9]

print("\ninsert(0, 0) method example:")
list1.insert(0, 0)  # This will insert the number 0 at index 0 of the list, resulting in [0, 1, 28, 33, 4, 3, 6, 7, 8, 9]
print(list1)  # This will print the updated list, which is [0, 1, 28, 33, 4, 3, 6, 7, 8, 9]

print("\nremove(28) method example:")
list1.remove(28)  # This will remove the first occurrence of the number 28 from the list, resulting in [0, 1, 33, 4, 3, 6, 7, 8, 9]
print(list1)  # This will print the updated list, which is [0, 1, 33, 4, 3, 6, 7, 8, 9]

print("\npop() method example:")
list1.pop()  # This will remove the last element from the list, resulting in [0, 1, 33, 4, 3, 6, 7, 8]
print(list1)  # This will print the updated list, which is [0, 1, 33, 4, 3, 6, 7, 8]

print("\nsort() method example:")
list1.sort()  # This will sort the list in ascending order, resulting in [0, 1, 3, 4, 6, 7, 8, 33]
print(list1)  # This will print the sorted list, which is [0, 1, 3, 4, 6, 7, 8, 33]

print("\nreverse() method example:")
list1.reverse()  # This will reverse the order of the list, resulting in [33, 8, 7, 6, 4, 3, 1, 0]
print(list1)  # This will print the reversed list, which is [33,_8,_7,_6,_4,_3,_1,_0]

print("\ncopy() method example:")
list2=list1.copy()  # This will create a shallow copy of the list, resulting in a new list that is identical to the original list
print(list2)  # This will print the copied list

print("\nlist slicing example:")
print(list1[0:3])  # This will print the first three elements of the list, which is [33, 8, 7]
print(list1[3:6])  # This will print the elements from index 3 to 5 of the list, which is [6, 4, 3]
print(list1[6:9])  # This will print the elements from index 6 to 8 of the list, which is [1, 0]

print("\nMemory address of list1 and list2:\n")
print(f'Memory address of list1: {id(list1)}')  # This will print the memory address of list1
print(f'Memory address of list2: {id(list2)}')  # This will print the memory address of list2, which is different from the memory address of list1, indicating that they are two separate lists
```

### 09_sets 

**Methods:**

- add()
- remove()
- union()
- intersection()
- difference()

**Use**

- Remove duplicates

```python
set1= {1, 2, 3, 4, 5}
print("\nSet Example:")
set1.add(6)  # This will add the number 6 to the set, resulting in {1, 2, 3, 4, 5, 6}
print(set1)  # This will print the updated set, which is {1, 2, 3, 4, 5, 6}

set1.remove(3)  # This will remove the number 3 from the set, resulting in {1, 2, 4, 5, 6}
print(set1)  # This will print the updated set, which is {1, 2, 4, 5, 6}

set1.union({7, 8, 9})  # This will return a new set that is the union of set1 and {7, 8, 9}, resulting in {1, 2, 4, 5, 6, 7, 8, 9}
print(set1)  # This will print the original set, which is still {1, 2, 4, 5, 6} because the union() method does not modify the original set

set1.intersection({4, 5, 6, 7, 8})  # This will return a new set that is the intersection of set1 and {4, 5, 6, 7, 8}, resulting in {4, 5, 6}
print(set1)  # This will print the original set, which is still {1, 2, 4, 5, 6} because the intersection() method does not modify the original set

set1.difference({4, 5, 6, 7, 8})  # This will return a new set that is the difference of set1 and {4, 5, 6, 7, 8}, resulting in {1, 2}
print(set1)  # This will print the original set, which is still {1, 2, 4, 5, 6} because the difference() method does not modify the original set
```

### 10_tuples

**Document**

- Immutable
- Faster than lists

```python
tuple1 = (1, 2, 3, 4, 5, 6)

print("\nTuple Example:")
print(tuple1)  # This will print the tuple, which is (1, 2, 3, 4, 5, 6)

print("\nAccessing Tuple Elements:")
print(tuple1[0])  # This will print the first element of the tuple, which is 1
print(tuple1[1])  # This will print the second element of the tuple, which is 2
print(tuple1[2])  # This will print the third element of the tuple, which is 3

print('\nAccessing Tuple Elements using Negative Indexing:')
print(tuple1[-1])  # This will print the last element of the tuple, which is 6
print(tuple1[-2])  # This will print the second to last element of the tuple, which is 5
print(tuple1[-3])  # This will print the third to last element of the tuple, which is 4


print("\nSlicing the Tuple:")
print(tuple1[0:3])  # This will print the first three elements of the tuple, which is (1, 2, 3)
print(tuple1[3:])   # This will print the elements from index 3 to the end of the tuple, which is (4, 5, 6)
print(tuple1[:3])   # This will print the first three elements of the tuple, which is (1, 2, 3)
print(tuple1[::2])  # This will print every second element of the tuple, which is (1, 3, 5)

print("\nTuple Unpacking:")
a, b, c, d, e, f = tuple1  # This will unpack the elements of the tuple into separate variables
print(a)  # This will print the value of variable a, which is 1
print(b)  # This will print the value of variable b, which is 2
print(c)  # This will print the value of variable c, which is 3
print(d)  # This will print the value of variable d, which is 4
print(e)  # This will print the value of variable e, which is 5
print(f)  # This will print the value of variable f, which is 6

print("\nWays to Add Values to a Tuple:")
# There will be a way to add values to tuple
#1. We can convert the tuple to a list, add the new value to the list, and then convert it back to a tuple.
tuple1 = tuple(list(tuple1) + [7])  # This will add the number 7 to the tuple, resulting in (1, 2, 3, 4, 5, 6, 7)
print(tuple1)  # This will print the updated tuple, which is (1, 2, 3, 4, 5, 6, 7)

#2. We can also use the + operator to concatenate the original tuple with a new tuple containing the new value.
tuple1 = tuple1 + (8,)  # This will add the number 8 to the tuple, resulting in (1, 2, 3, 4, 5, 6, 7, 8)
print(tuple1)  # This will print the updated tuple, which is (1, 2, 3, 4, 5, 6, 7, 8)

```
