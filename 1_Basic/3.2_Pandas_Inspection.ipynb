```python
from datasets import load_dataset
import pandas as pd

dataset = load_dataset('lukebarousse/data_jobs')
raw=dataset['train'].to_pandas()
df = None
df = raw.copy()
# Dataset 
```

```python
df['job_posted_date']=pd.to_datetime(df.job_posted_date)
```

**tail() Method**

```python
# bottom 10 records
df.tail(10)
```

**iloc[] Method**

```python
# Looking between 50 and 60, there is a method, that is iloc[]
df.iloc[50:60]
```

**Accessing dataset**

```python
df['job_title_short'].head(10) # or use tail() to see bottom records
```

**Accessing multiple columns**

```python
df[['job_title_short','salary_year_avg']]
```

**Accessing multiple columns & records using iloc[] Method**

```python
df.iloc[50:60,2:5]
# In this method, there is a disadvantage, which is that we can’t select the column we want
```

```python
# To over come that iloc[] methods disadvantage 
raw[['job_title_short','search_location']].iloc[50:60]
```

**filtering the dataset.**

```python
df[df.job_title_short == 'Data Analyst']
```

**Multiple filters**


```python
df[(df.job_title_short =='Data Analyst') & (df.salary_year_avg>10000)]
```

**Multiple Filter with notna() Method**

```python
df[(df.job_title_short=='Data Analyst')&(df.salary_year_avg.notna())] #notna() method is used to remove the NaN values 
```
