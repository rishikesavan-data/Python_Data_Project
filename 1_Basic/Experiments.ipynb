```python
class calc:
    def __init__(self):
        self.a = 0
        self.b = 0

    def inputs(self):
        self.a = int(input('enter the 1st number: '))
        self.b = int(input("enter the 2nd number: "))

    def display_option(self):
        print("press 1 to do add")
        print("press 2 to do subtract")
        print("press 3 to do multiply")
        print("press 4 to do divide")
        print("press 5 to do modulus")
        print("press 6 to do power")
        print("press 7 to do floor division")
        print("press 8 to do exit")
        print("--------------------------------\n")
        
    def solve(self):
        self.inputs()
        self.display_option()
        option=int(input("enter your option: "))
        if option==1:
            print(self.a+self.b)
        elif option==2:
            print(self.a-self.b)
        elif option==3:
            print(self.a*self.b)
        elif option==4:
            print(self.a/self.b)
        elif option==5:
            print(self.a%self.b)
        elif option==6:
            print(self.a**self.b)
        elif option==7:
            print(self.a//self.b)
        elif option==8:
            exit()

c=calc()
choice = input("do you want to continue? (y/n): ")
while choice == "y":
    c.solve()
    choice = input("do you want to continue? (y/n): ")
```
