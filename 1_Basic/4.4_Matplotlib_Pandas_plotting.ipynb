# Matplotlib Pandas: Ploting

*In this section we are going to use pandas for plot graph using pd.plot() method* 

*maybe you may find this method is easier than the previous methods*

```python
from datasets import load_dataset
import pandas as pd
import matplotlib.pyplot as plt

raw = load_dataset('lukebarousse/data_jobs') 
df = raw.copy()
df =df['train'].to_pandas()
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
df['job_posted_month'] = df.job_posted_date.dt.month
df.info()
```

```python
jobs=df.job_title_short.value_counts()
jobs
```

```python
jobs.plot(kind='bar')
plt.title('Job_Postings')
plt.xlabel('')
plt.ylabel('job_count')
plt.xticks(rotation = 45,ha='right')
plt.show()
```

```python
jobs.plot(kind='line')
plt.title('Job_Postings')
plt.xlabel('')
plt.ylabel('job_count')
plt.xticks(rotation = 45,ha='right')
plt.show()
```

```python
df.plot(x='job_title_short',y='salary_year_avg',kind='line')
plt.xlabel('')
plt.xticks(rotation = 45,ha='right')
plt.show()
```

# Excersise: Matplotlib Basics

```python
median_salary=df.groupby('job_title_short')['salary_year_avg'].median().sort_values()
median_salary
```

```python
median_salary.plot(kind='barh')
plt.ylabel('')
plt.xlabel('Salary in $(USD)')
plt.title('Job Salary ')
```
