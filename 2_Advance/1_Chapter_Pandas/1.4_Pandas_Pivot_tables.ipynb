# Pandas: Pivot_Tables
**Method**
- df.pivot_table()

```python
# Import libraries

from datasets import load_dataset
import pandas as pd
import matplotlib.pyplot as plt

# Load Dataset
raw = load_dataset('lukebarousse/data_jobs')
df = raw['train'].to_pandas()

# Fixing date 
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
df
```

**pivot table**

*parameters example*
- index
- aggfunc

```python
df.pivot_table(index='job_title_short',aggfunc='size')
```

```python
df.groupby('job_title_short').size()
```

*parameter example*
- values 

```python
df.pivot_table(values='salary_year_avg',index='job_title_short',aggfunc='median')
```

```python
df.groupby('job_title_short')['salary_year_avg'].median()
```

*parameter example*
- columns 

```python
df.pivot_table(values='salary_year_avg',index='job_title_short',columns='job_country',aggfunc='median')
# the table looks lengthy, so i interchange the index and columns to make it look good.
```

*Index and columns parameter values interchange example*

```python
df.pivot_table(values='salary_year_avg',index='job_country',columns='job_title_short',aggfunc='median')
```

```python
top_countries = df.job_country.value_counts().head(7).index
top_countries

df_pivot_table = df.pivot_table(
    values= 'salary_year_avg',
    index= 'job_country',
    columns='job_title_short',
    aggfunc='median'
)
job_titles =['Data Analyst','Data Engineer','Data Scientist']

df_pivot_table = df_pivot_table.loc[top_countries,job_titles]
df_pivot_table.plot(kind='bar')
plt.xlabel('Country')
plt.xticks(rotation=45,ha='right')
plt.ylabel('Median Yearly Salary $(USD)')
plt.title('Median Salary by Country and Job')
plt.legend()
plt.show()
```
