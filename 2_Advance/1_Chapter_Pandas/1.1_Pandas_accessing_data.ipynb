# Pandas_Accessing_data

**Method**
- df.loc[]

```python
from datasets import load_dataset
import pandas as pd

raw =load_dataset('lukebarousse/data_jobs')
df = raw.copy()
df = df['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
df.info()
```

**df.loc[] Method**


```python
df.iloc[10:25,1:5] # In this method we use to slice the column using numbers 1:5 , after slicing the rows 10:20
# what if the i don't know the column index 
```

**df.loc[] Method**

```python
df.loc[:,'salary_rate':'salary_hour_avg'].dropna(subset='salary_rate')
```
