**df.copy() Method**

**NOTE:**
- This copy method is used to copy the data. Why do we need to use this when we have a '=' assignment operator
- This assignment operator will modify the raw dataset also. for example a = b.
- if we modify the b, a also gets modified when using the assignment operator
- To avoid the original records from being modified, we use the copy() method 

```python
df_original = df.copy() ## Here i make a copy of df and store it in the df_original 
## if i make any change in the df_original, it will not change anything in the df. 

df_copy = df_original ## Here i use assignment operator '=' to assign the df_original to df_copy 
## if i make any change in the df_copy, it will also reflect in the df_original 

print(df_copy.loc[:,'salary_rate':'salary_hour_avg'].dropna(subset=['salary_rate']))

print()
print(id(df))
print(id(df_copy))
print(id(df)==id(df_copy))

# both id's are same if we modify the df_copy, df also get modify. This is a major problem, we lost the original dataset. 
# so to protect the original dataset, we need to use this copy() method
```

**Demonstration**

```python
median_yearly_salary = df.salary_year_avg.median()
print('median_yearly_salary: ',median_yearly_salary)

median_hourly_salary = df.salary_hour_avg.median()
print('median_hourly_salary: ',median_hourly_salary)
```

```python
df_copy['salary_year_avg'] = df_copy['salary_year_avg'].fillna(median_yearly_salary)
df_copy['salary_hour_avg'] = df_copy['salary_hour_avg'].fillna(median_hourly_salary)

df_copy.loc[:,'salary_rate':'salary_hour_avg'].dropna()
```

```python
df_original.loc[:,'salary_rate':'salary_hour_avg'].dropna(subset='salary_rate')
# Look i modify the df_copy but the df_original also get modified
```
