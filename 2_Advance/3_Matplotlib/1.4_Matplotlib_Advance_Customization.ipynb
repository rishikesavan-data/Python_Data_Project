# Matplotlib : Advance Customization

```python
# Import Libraries
from datasets import load_dataset
import pandas as pd
import ast
import matplotlib.pyplot as plt
from adjustText import adjust_text

# Load Dataset
raw = load_dataset('lukebarousse/data_jobs')
data = raw['train'].to_pandas()
df = data.copy()

# Data Cleaning 
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
df
```

**Example 1**

*Filtering the dataset for only Data Analyst role and then adding the month number column*
- copy()
- dt.month

```python
df_DA =df[df['job_title_short']=='Data Analyst'].copy()
```

```python
df_DA['job_posted_month'] = df['job_posted_date'].dt.month
df_DA
```

*unpacking the job_skills column using explode() Method*
- explode()

```python
df_DA_explode = df_DA.explode('job_skills')
df_DA_explode
```

*To find the in-demand skill we use pivot_table() method*
- Pivot_table()

```python
df_DA_pivot =df_DA_explode.pivot_table(index='job_posted_month',columns='job_skills',aggfunc='size',fill_value=0)
df_DA_pivot
```

*Adding the total to find the top in-demand skills*
- df.loc[]
- df.sum()

```python
df_DA_pivot.loc['total'] = df_DA_pivot.sum()
df_DA_pivot
```

*sorting the table*
- sort_values()

```python
df_DA_pivot_sorted =df_DA_pivot[df_DA_pivot.loc['total'].sort_values(ascending=False).index]
df_DA_pivot_sorted
```

*Removing the total why because it's distrub the ploting*
- drop()

```python
df_DA_pivot_sorted.drop('total',inplace=True)
```

```python
df_DA_pivot_sorted
```

*Reset the index to change the job_posted_month column from number '1' format to text 'Jan'*
- reset_index()
- apply()
- set_index()
- lambda function

```python
df_DA_pivot_sorted.reset_index(inplace=True)
```

```python
df_DA_pivot_sorted['job_posted_month'] = df_DA_pivot_sorted['job_posted_month'].apply(lambda x : pd.to_datetime(x, format='%m').strftime('%b'))
df_DA_pivot_sorted.set_index('job_posted_month',inplace= True)
```

> *Advance Customization Example 1*


```python
df_DA_pivot_sorted.iloc[:,:5].plot(
    kind='line',
    linewidth=3,
    linestyle = '-',
    marker='o',
    markersize =7,
    colormap='viridis',
    figsize=(10,5))
plt.title('Top 5 In-Demand Skills for Data Analyst Per Month')
plt.xlabel('Month')
plt.ylabel('Skill Count')
plt.legend()
plt.tight_layout()
plt.show()
```

*plotting the name of all the months in the x-axis*

```python
ax = df_DA_pivot_sorted.iloc[:, :5].plot(
    kind='line',
    linewidth=3,
    linestyle='-',
    marker='o',
    markersize=7,
    colormap='viridis',
    figsize=(10, 5)
)

ax.set_xticks(range(len(df_DA_pivot_sorted.index)))
ax.set_xticklabels(df_DA_pivot_sorted.index)

plt.title('Top 5 In-Demand Skills for Data Analyst Per Month')
plt.xlabel('Month')
plt.ylabel('Skill Count')
plt.legend()
plt.tight_layout()
plt.show()
```

**Example 2**

```python
top_skills =20
df_groupby =df_DA_explode.groupby('job_skills').agg(skill_count=('job_skills','size'),salary =('salary_year_avg','median')).sort_values('skill_count',ascending=False).head(top_skills)
df_groupby
```

> *Advance Customization Example 2*

```python
df_groupby.plot(kind='scatter',x='skill_count',y='salary',marker='D',figsize=(12,7))

plt.xlabel('Count')
plt.ylabel('Median Salary ($USD)')
plt.title('Salary Vs. Count for the Top 10 Skills of a Data Analyst')

text =[]
for i,txt in enumerate(df_groupby.index):
    text.append(plt.text(df_groupby['skill_count'].iloc[i],df_groupby['salary'].iloc[i],txt))
adjust_text(text,arrowprops=dict(arrowstyle='->',color='grey',lw=1))

ax=plt.gca()
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda y,pos :f'${int(y/1000)}K'))
plt.show()
```
