# Matplotlib : Pie Plot

**Methods**
- df.plot(kind='pie') or plt.pie()
- fig, ax = plt.subplots()

```python
# Import Libraries
from datasets import load_dataset
import pandas as pd
import ast
import matplotlib.pyplot as plt

# Load_dataset
raw = load_dataset('lukebarousse/data_jobs')
data = raw['train'].to_pandas()
df = data.copy()

# Data Cleaning
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
df['job_skills'] = df['job_skills'].apply(lambda x : ast.literal_eval(x) if pd.notna(x) else x)

df
```

*Filtering out only the true, false column*

```python
df_filter = df[['job_work_from_home','job_no_degree_mention','job_health_insurance']]
df_filter
```

**Method**
- df.plot(kind='pie')

```python
df_filter['job_work_from_home'].value_counts().plot(kind='pie',startangle = 90,autopct='%1.2f%%')
plt.title('Work from home',fontsize=15)
```

```python
df_filter['job_health_insurance'].value_counts().plot(kind='pie',startangle=90,autopct='%1.2f%%')
plt.title('Health Insurance',fontsize=15)
```

```python
df_filter['job_no_degree_mention'].value_counts().plot(kind='pie',startangle=90,autopct='%1.2f%%')
plt.title('No Degree Mention',fontsize=15)
```

**Method**
- fig, ax = plt.subplots()
- for loops with enumerate and items methods.

```python
fig, ax = plt.subplots(1,3)

dict_lib ={'job_work_from_home':'Work From Home',
           'job_health_insurance':'Health Insurance',
           'job_no_degree_mention':'No Degree Mention'}

for i,(column,title) in enumerate(dict_lib.items()):
    df_filter[column].value_counts().plot(kind='pie',startangle = 90,autopct = '%1.1f%%',ax=ax[i],title=title)
```
