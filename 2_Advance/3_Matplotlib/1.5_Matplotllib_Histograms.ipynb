# Matplotlib : Histograms

```python
# Import Libraries
from datasets import load_dataset
import pandas as pd
import ast
import matplotlib.pyplot as plt

# Load Dataset
raw = load_dataset('lukebarousse/data_jobs')
data = raw['train'].to_pandas()
df = data.copy()

# Data Cleaning 
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
df['job_skills'] = df['job_skills'].apply(lambda x : ast.literal_eval(x) if pd.notna(x) else(x))
```

```python
df
```

**Filtering the dataset**

```python
df_filter= df[(df['job_title_short']=='Data Analyst') & (df['job_country']=='United States')]
```

```python
df_filter
```

**Plotting Histogram**

```python
df_filter['salary_year_avg'].plot(kind='hist',bins=40,edgecolor='black')
plt.title('Distibution of United States Yearly Salary for Data Analyst')
plt.xlabel('Yearly Salary')
plt.ylabel('Number of Jobs')
plt.xlim(0,250000)
ax = plt.gca()
ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x,pos:f'${int(x/1000)}K'))
```
