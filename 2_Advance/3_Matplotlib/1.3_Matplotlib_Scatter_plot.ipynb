# Matplotlib : Scatter Plot

**method**
- df.plot(kind='scatter')


```python
# Import Libraries
from datasets import load_dataset
import pandas as pd
import ast
import matplotlib.pyplot as plt

# Load Dataset
raw = load_dataset('lukebarousse/data_jobs')
data = raw['train'].to_pandas()
df = data.copy()

# Data Cleaning 
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
df['job_skills'] = df['job_skills'].apply(lambda x : ast.literal_eval(x) if pd.notna(x) else x)

df
```

**Example 1**

```python
dict_skills ={'skills':['python','sql','excel','power bi'],
'count' :[91,100,78,70],
'salary':[181,200,150,160]
}

datas = pd.DataFrame(dict_skills)
datas
```

```python
datas.plot(kind='scatter',x='count',y='salary')
plt.xlabel('Job Count')
plt.ylabel('Salary')
plt.title('salary vs. count by job skills')
```

*Using explode method to unpack the job_skills*

```python
df_explode = df.explode('job_skills')
```

```python
df_explode
```

**Example 2** 

*Filtering out only the data analyst role and group by skill_count and median salary with skill name*

```python
df_Da= df_explode[df_explode['job_title_short']=='Data Analyst']
df_group =df_Da.groupby('job_skills').agg(skill_count=('job_skills','size'),salary = ('salary_year_avg','median'))
df_group 
```

*sorting the values of the skill count in descending order and filtering out only the top 10*

```python
df_skills =df_group.sort_values('skill_count',ascending=False).head(10)
df_skills
```

**Method**
- df.plot(kind='scatter')

*ploting the skill stats the we found*

```python
df_skills.plot(kind='scatter',x='skill_count',y='salary')
plt.xlabel('Count')
plt.ylabel('Median Salary ($USD)')
plt.title('Salary Vs. Count for the Top 10 Skills of a Data Analyst')
```

**Method**
- plt.text()

*Naming the points using for loop and plt.text() method*

```python
df_skills.plot(kind='scatter',x='skill_count',y='salary')
plt.xlabel('Count')
plt.ylabel('Median Salary ($USD)')
plt.title('Salary Vs. Count for the Top 10 Skills of a Data Analyst')

for i,txt in enumerate(df_skills.index):
    plt.text(df_skills['skill_count'].iloc[i],df_skills['salary'].iloc[i],txt)

```
