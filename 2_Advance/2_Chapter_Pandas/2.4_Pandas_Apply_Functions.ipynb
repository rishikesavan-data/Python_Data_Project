```python
# Import libraries 
from datasets import load_dataset
import pandas as pd
import ast

# load dataset
raw=load_dataset('lukebarousse/data_jobs')
data = raw['train'].to_pandas()
df = data.copy()
df_job = data.copy()

# String to Date Conversion 
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
df_job['job_posted_date'] = pd.to_datetime(df_job.job_posted_date)
print(df.info())
df
```

**2. The challenge is to change the job_skill column into list type from string**

```python
print(type(df['job_skills'][8]))
df['job_skills'][8]
```

```python
print(type(ast.literal_eval(df.job_skills[7])))
ast.literal_eval(df.job_skills[7])
```

*1. Method*

```python
def clean_list(skill_list):
    if pd.notna(skill_list):
        return ast.literal_eval(skill_list)

df['job_skills'] = df['job_skills'].apply(clean_list)
```

*2. Method*

```python
df_job['job_skills'] = df_job['job_skills'].apply(lambda skill : ast.literal_eval(skill) if pd.notna(skill) else skill)
```

**3. Calculates Projected salary next year**
- Senior roles assume 5%
- other roles assume 3%

*1. Method*

```python
def projected_salary(row):
    if 'Senior' in row['job_title_short']:
        return row['salary_year_avg']*1.05
    else:
        return row['salary_year_avg']*1.03
    
df_salary['salary_inflated'] = df_salary.apply(projected_salary,axis =1)
df_salary[['job_title_short','salary_year_avg','salary_inflated']].tail(10)
```

```python
df_salary.drop('salary_inflated',axis=1,inplace = True)
```

*2. Method*

```python
df_salary['salary_inflated'] = df_salary.apply(lambda row: row['salary_year_avg']*1.05 if 'Senior' in row['job_title_short'] else row['salary_year_avg']*1.03,axis =1 )
```
